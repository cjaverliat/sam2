# Object masks in images from prompts with EfficientTAM


[EfficientTAM](https://github.com/yformer/EfficientTAM) predicts object masks given prompts that indicate the desired object, at 18M parameters against SAM 2 large's 224M. It converts the image into an image embedding that allows high quality masks to be efficiently produced from a prompt.

It loads through the SAM 2 builder and answers the SAM 2 API, so this is [`sam2_image_predictor_example.ipynb`](./sam2_image_predictor_example.ipynb) with one config and one checkpoint changed.

The generic predictor (`build_sam2_predictor`) exposes this directly: `encode_image` computes the image embeddings once, `encode_prompts` encodes point/box/mask prompts, and `generate_masks` predicts masks from those prompts. The model can take as input both point and box prompts, as well as masks from a previous iteration of prediction. The thin `sam2_predict` helper defined below wraps these calls for a single image and object.

```python
np.random.seed(3)

def show_mask(mask, ax, random_color=False, borders = True):
    if random_color:
        color = np.concatenate([np.random.random(3), np.array([0.6])], axis=0)
    else:
        color = np.array([30/255, 144/255, 255/255, 0.6])
    h, w = mask.shape[-2:]
    mask = mask.astype(np.uint8)
    mask_image =  mask.reshape(h, w, 1) * color.reshape(1, 1, -1)
    if borders:
        import cv2
        contours, _ = cv2.findContours(mask,cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE) 
        # Try to smooth contours
        contours = [cv2.approxPolyDP(contour, epsilon=0.01, closed=True) for contour in contours]
        mask_image = cv2.drawContours(mask_image, contours, -1, (1, 1, 1, 0.5), thickness=2) 
    ax.imshow(mask_image)

def show_points(coords, labels, ax, marker_size=375):
    pos_points = coords[labels==1]
    neg_points = coords[labels==0]
    ax.scatter(pos_points[:, 0], pos_points[:, 1], color='green', marker='*', s=marker_size, edgecolor='white', linewidth=1.25)
    ax.scatter(neg_points[:, 0], neg_points[:, 1], color='red', marker='*', s=marker_size, edgecolor='white', linewidth=1.25)   

def show_box(box, ax):
    x0, y0 = box[0], box[1]
    w, h = box[2] - box[0], box[3] - box[1]
    ax.add_patch(plt.Rectangle((x0, y0), w, h, edgecolor='green', facecolor=(0, 0, 0, 0), lw=2))    

def show_masks(image, masks, scores, point_coords=None, box_coords=None, input_labels=None, borders=True):
    for i, (mask, score) in enumerate(zip(masks, scores)):
        plt.figure(figsize=(10, 10))
        plt.imshow(image)
        show_mask(mask, plt.gca(), borders=borders)
        if point_coords is not None:
            assert input_labels is not None
            show_points(point_coords, input_labels, plt.gca())
        if box_coords is not None:
            # boxes
            show_box(box_coords, plt.gca())
        if len(scores) > 1:
            plt.title(f"Mask {i+1}, Score: {score:.3f}", fontsize=18)
        plt.axis('off')
        plt.show()
```

## Selecting objects with SAM 2

First, load the model and predictor. Change the path below to point to the EfficientTAM checkpoint — `ti` is the smallest, `s` the other size. Running on CUDA is recommended.

```python
from sam.build_sam import build_sam2_predictor

sam2_checkpoint = "../checkpoints/efficienttam_ti.pt"
model_cfg = "configs/efficienttam/efficienttam_ti.yaml"

model = build_sam2_predictor(model_cfg, sam2_checkpoint, device=device, apply_postprocessing=True, use_half=True)


def encode_image(model, image):
    """Encode an HWC uint8 image into multi-scale image embeddings."""
    img = torch.as_tensor(image, device=model.device).permute(2, 0, 1)
    return model.encode_image(img)


def sam2_predict(
    model,
    img_embeddings,
    orig_hw,
    point_coords=None,
    point_labels=None,
    box=None,
    mask_input=None,
    multimask_output=True,
):
    """Predict masks for a single image/object using the generic SAM 2 API.

    Returns (masks, scores, logits) as numpy arrays sorted by the model's
    prompt order. `logits` are full-resolution mask logits that can be passed
    back as `mask_input` to refine a prediction.
    """
    device = model.device
    point_coords_t = point_labels_t = boxes_t = masks_logits_t = None
    if point_coords is not None:
        point_coords_t = torch.as_tensor(
            point_coords, dtype=torch.float32, device=device
        ).reshape(1, -1, 2)
        point_labels_t = torch.as_tensor(
            point_labels, dtype=torch.int, device=device
        ).reshape(1, -1)
    if box is not None:
        boxes_t = torch.as_tensor(box, dtype=torch.float32, device=device).reshape(
            1, -1, 4
        )
    if mask_input is not None:
        mask_input_t = torch.as_tensor(mask_input, dtype=torch.float32, device=device)
        masks_logits_t = mask_input_t.reshape(1, 1, *mask_input_t.shape[-2:])

    prompt_embeddings = model.encode_prompts(
        orig_hw=orig_hw,
        batch_size=1,
        points_coords=point_coords_t,
        points_labels=point_labels_t,
        boxes=boxes_t,
        masks_logits=masks_logits_t,
    )
    result = model.generate_masks(
        orig_hw=orig_hw,
        img_embeddings=img_embeddings,
        prompt_embeddings=prompt_embeddings,
        multimask_output=multimask_output,
    )
    masks = (result.masks_logits[0] > model.mask_threshold).cpu().numpy()
    scores = result.ious[0].float().cpu().numpy()
    logits = result.masks_logits[0].float().cpu().numpy()
    return masks, scores, logits
```

Process the image to produce an image embedding by calling `encode_image`. We keep this embedding (and the original image size `orig_hw`) and reuse them for subsequent mask predictions.

```python
img_embeddings, _ = encode_image(model, image)
orig_hw = image.shape[:2]
```

To select the truck, choose a point on it. Points are input to the model in (x,y) format and come with labels 1 (foreground point) or 0 (background point). Multiple points can be input; here we use only one. The chosen point will be shown as a star on the image.

```python
input_point = np.array([[500, 375]])
input_label = np.array([1])
```

```python
plt.figure(figsize=(10, 10))
plt.imshow(image)
show_points(input_point, input_label, plt.gca())
plt.axis('on')
plt.show()  
```

```python
print(img_embeddings[-1].shape)  # lowest-resolution image embedding
```

Predict with the `sam2_predict` helper. The model returns masks, quality predictions for those masks, and (full-resolution) mask logits that can be passed to the next iteration of prediction as `mask_input`.

```python
masks, scores, logits = sam2_predict(
    model,
    img_embeddings,
    orig_hw,
    point_coords=input_point,
    point_labels=input_label,
    multimask_output=True,
)
sorted_ind = np.argsort(scores)[::-1]
masks = masks[sorted_ind]
scores = scores[sorted_ind]
logits = logits[sorted_ind]
```

With `multimask_output=True` (the default setting), SAM 2 outputs 3 masks, where `scores` gives the model's own estimation of the quality of these masks. This setting is intended for ambiguous input prompts, and helps the model disambiguate different objects consistent with the prompt. When `False`, it will return a single mask. For ambiguous prompts such as a single point, it is recommended to use `multimask_output=True` even if only a single mask is desired; the best single mask can be chosen by picking the one with the highest score returned in `scores`. This will often result in a better mask.

```python
masks.shape  # (number_of_masks) x H x W
```

```python
show_masks(image, masks, scores, point_coords=input_point, input_labels=input_label, borders=True)
```

## Specifying a specific object with additional points

The single input point is ambiguous, and the model has returned multiple objects consistent with it. To obtain a single object, multiple points can be provided. If available, a mask from a previous iteration can also be supplied to the model to aid in prediction. When specifying a single object with multiple prompts, a single mask can be requested by setting `multimask_output=False`.

```python
input_point = np.array([[500, 375], [1125, 625]])
input_label = np.array([1, 1])

mask_input = logits[np.argmax(scores), :, :]  # Choose the model's best mask
```

```python
masks, scores, _ = sam2_predict(
    model,
    img_embeddings,
    orig_hw,
    point_coords=input_point,
    point_labels=input_label,
    mask_input=mask_input,
    multimask_output=False,
)
```

```python
show_masks(image, masks, scores, point_coords=input_point, input_labels=input_label)
```

## Batched prompt inputs

The generic API can take multiple input prompts for the same image by batching them: encode the prompts with `batch_size=B`, broadcast the image embeddings over that batch, and call `generate_masks`. For example, imagine we have several box outputs from an object detector.

```python
input_boxes = np.array([
    [75, 275, 1725, 850],
    [425, 600, 700, 875],
    [1375, 550, 1650, 800],
    [1240, 675, 1400, 750],
])
```

```python
# Each box is its own object, so we build a batch of single-box prompts and
# broadcast the (single) image embeddings over that batch.
boxes_t = torch.as_tensor(input_boxes, dtype=torch.float32, device=device)[:, None, :]
batch_size = boxes_t.shape[0]

prompt_embeddings = model.encode_prompts(
    orig_hw=orig_hw, batch_size=batch_size, boxes=boxes_t
)
img_embeddings_b = [e.expand(batch_size, *e.shape[1:]) for e in img_embeddings]
result = model.generate_masks(
    orig_hw=orig_hw,
    img_embeddings=img_embeddings_b,
    prompt_embeddings=prompt_embeddings,
    multimask_output=False,
)
masks = (result.masks_logits > model.mask_threshold).cpu().numpy()
```

```python
masks.shape  # (batch_size) x (num_predicted_masks_per_input) x H x W
```

```python
plt.figure(figsize=(10, 10))
plt.imshow(image)
for mask in masks:
    show_mask(mask.squeeze(0), plt.gca(), random_color=True)
for box in input_boxes:
    show_box(box, plt.gca())
plt.axis('off')
plt.show()
```

## End-to-end batched inference
If all prompts are available in advance, it is possible to run SAM 2 directly in an end-to-end fashion. The generic API operates on a single image resolution per call, so we loop over the images and run each image's prompts in one batched call.

```python
image1 = image  # truck.jpg from above
image1_boxes = np.array([
    [75, 275, 1725, 850],
    [425, 600, 700, 875],
    [1375, 550, 1650, 800],
    [1240, 675, 1400, 750],
])

image2 = Image.open('images/groceries.jpg')
image2 = np.array(image2.convert("RGB"))
image2_boxes = np.array([
    [450, 170, 520, 350],
    [350, 190, 450, 350],
    [500, 170, 580, 350],
    [580, 170, 640, 350],
])

img_batch = [image1, image2]
boxes_batch = [image1_boxes, image2_boxes]
```

```python
# The generic API operates on one image resolution per call, so we encode each
# image and run all of its prompts in a single batched call, looping over images.
```

```python
masks_batch = []
for img, boxes in zip(img_batch, boxes_batch):
    img_embeddings_i, _ = encode_image(model, img)
    orig_hw_i = img.shape[:2]
    boxes_t = torch.as_tensor(boxes, dtype=torch.float32, device=device)[:, None, :]
    batch_size = boxes_t.shape[0]

    prompt_embeddings = model.encode_prompts(
        orig_hw=orig_hw_i, batch_size=batch_size, boxes=boxes_t
    )
    img_embeddings_b = [e.expand(batch_size, *e.shape[1:]) for e in img_embeddings_i]
    result = model.generate_masks(
        orig_hw=orig_hw_i,
        img_embeddings=img_embeddings_b,
        prompt_embeddings=prompt_embeddings,
        multimask_output=False,
    )
    masks_batch.append((result.masks_logits > model.mask_threshold).cpu().numpy())
```

```python
for image, boxes, masks in zip(img_batch, boxes_batch, masks_batch):
    plt.figure(figsize=(10, 10))
    plt.imshow(image)   
    for mask in masks:
        show_mask(mask.squeeze(0), plt.gca(), random_color=True)
    for box in boxes:
        show_box(box, plt.gca())
```

Similarly, we can have a batch of point prompts defined over a batch of images

```python
image1 = image  # truck.jpg from above
image1_pts = np.array([
    [[500, 375]],
    [[650, 750]]
    ]) # Bx1x2 where B corresponds to number of objects 
image1_labels = np.array([[1], [1]])

image2_pts = np.array([
    [[400, 300]],
    [[630, 300]],
])
image2_labels = np.array([[1], [1]])

pts_batch = [image1_pts, image2_pts]
labels_batch = [image1_labels, image2_labels]
```

```python
masks_batch = []
scores_batch = []
for img, pts, labels in zip(img_batch, pts_batch, labels_batch):
    img_embeddings_i, _ = encode_image(model, img)
    orig_hw_i = img.shape[:2]
    pts_t = torch.as_tensor(pts, dtype=torch.float32, device=device)  # (B, 1, 2)
    labels_t = torch.as_tensor(labels, dtype=torch.int, device=device)  # (B, 1)
    batch_size = pts_t.shape[0]

    prompt_embeddings = model.encode_prompts(
        orig_hw=orig_hw_i,
        batch_size=batch_size,
        points_coords=pts_t,
        points_labels=labels_t,
    )
    img_embeddings_b = [e.expand(batch_size, *e.shape[1:]) for e in img_embeddings_i]
    result = model.generate_masks(
        orig_hw=orig_hw_i,
        img_embeddings=img_embeddings_b,
        prompt_embeddings=prompt_embeddings,
        multimask_output=True,
    )
    masks_batch.append((result.masks_logits > model.mask_threshold).cpu().numpy())
    scores_batch.append(result.ious.float().cpu().numpy())

# Select the best single mask per object
best_masks = []
for masks, scores in zip(masks_batch, scores_batch):
    best_masks.append(masks[range(len(masks)), np.argmax(scores, axis=-1)])
```

```python
for image, points, labels, masks in zip(img_batch, pts_batch, labels_batch, best_masks):
    plt.figure(figsize=(10, 10))
    plt.imshow(image)   
    for mask in masks:
        show_mask(mask, plt.gca(), random_color=True)
    show_points(points, labels, plt.gca())
```
